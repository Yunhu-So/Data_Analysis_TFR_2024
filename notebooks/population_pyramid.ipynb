```python
from pathlib import Path
import pandas as pd

def find_root(start: Path = Path.cwd()) -> Path:
    # Look for a directory that has 'data' and 'reports', walk up from current working dir
    for p in [start, *start.parents]:
        if (p / "data").exists() and (p / "reports").exists():
            return p
    return start

ROOT = find_root()
DATA_DIR = ROOT / "data" / "raw"
FIG_DIR  = ROOT / "reports" / "figures"
FIG_DIR.mkdir(parents=True, exist_ok=True)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Retrieving male population dataset (year 2008)
df_male = pd.read_excel(DATA_DIR / "Total Population in 2008 All Age Range.xlsx", skiprows=3, index_col='Province', usecols='B,E:Y', nrows=1)

df_male
```

```python
# Renaming column names
headerNames = ['0~4', '5~9', '10~14', '15~19', '20~24', '25~29', '30~34', '35~39', '40~44', '45~49', '50~54', '55~59', '60~64', '65~69', '70~74', '75~79', '80~84', '85~89', '90~94', '95~99', 'Above 100']
df_male.columns = headerNames

df_male
```

```python
 # Converting str -> int
df_male_temp1 = df_male.iloc[0:1, 0:-1]
df_male_temp1.iloc[0] = df_male_temp1.iloc[0].str.replace(',', '').astype(int)

# Combining two data frames
df_male_temp2 = df_male.iloc[0:1, -1:]
df_male = pd.concat([df_male_temp1,df_male_temp2], axis=1, join='inner')

df_male
```

```python
# Retrieving female population dataset
df_female = pd.read_excel(DATA_DIR / "Total Population in 2008 All Age Range.xlsx", skiprows=3, index_col='Province', usecols='B,AB:AV', nrows=1)

df_female
```

```python
# To have the same column names
df_female.columns = df_male.columns

# Converting str -> int
df_female.iloc[0] = df_female.iloc[0].str.replace(',', '').astype(int)

df_female
```

```python
# Creating population pyramid
plt.figure(figsize=(10, 8))
plt.barh(df_male.columns, -df_male.iloc[0] // 1000)
plt.barh(df_female.columns, df_female.iloc[0] // 1000)
plt.title("South Korea's Population Pyramid in 2008")
plt.xlabel("Population (in Thousands)")
plt.ylabel("Age Range")
plt.show()
```

```python
# Following the same step (year 2024)
df_male = pd.read_excel(DATA_DIR / "Total Population in 2024 All Age Range.xlsx", skiprows=3, index_col='Province', usecols='B,E:Y', nrows=1)

headerNames = ['0~4', '5~9', '10~14', '15~19', '20~24', '25~29', '30~34', '35~39', '40~44', '45~49', '50~54', '55~59', '60~64', '65~69', '70~74', '75~79', '80~84', '85~89', '90~94', '95~99', 'Above 100']
df_male.columns = headerNames
df_male.iloc[0] = df_male.iloc[0].str.replace(',', '').astype(int)

df_male
```

```python
df_female = pd.read_excel(DATA_DIR / "Total Population in 2024 All Age Range.xlsx", skiprows=3, index_col='Province', usecols='B,AB:AV', nrows=1)

df_female.columns = df_male.columns
df_female.iloc[0] = df_female.iloc[0].str.replace(',', '').astype(int)

df_female
```

```python
plt.figure(figsize=(10, 8))
plt.barh(df_male.columns, -df_male.iloc[0] // 1000)
plt.barh(df_female.columns, df_female.iloc[0] // 1000)
plt.title("South Korea's Population Pyramid in 2024")
plt.xlabel("Population (in Thousands)")
plt.ylabel("Age Range")
plt.show()
```
